# Implementing AI Guardrails

A hands-on, interactive course for AI engineers on defending LLM applications with runtime validation.

Every concept is paired with runnable code. You will build a working input/output guard pipeline and
watch real attacks succeed against an unguarded agent, then fail against a guarded one.

---

## What This Course Is About

An LLM application has a property no traditional web app has: **the instructions and the data are the
same thing.** Your system prompt, the user's message, a retrieved document, and a tool's response all
become one undifferentiated stream of tokens.

That single fact is why prompt injection is not a bug you patch, and why *asking the model nicely* is
not a security control. As the [OWASP GenAI LLM Top 10](https://genai.owasp.org/resource/owasp-genai-llm-top-10-2026/)
puts it:

> *"LLMs make no architectural distinction between 'instructions' and 'data' (both are tokens on the
> same stream), so there is no clean equivalent to parameterized queries."*

The course thesis follows directly from that: **defense is architectural, not interceptive.** Assume the
instruction boundary will eventually be bypassed, then constrain what the model can do and what its
output can reach. Guardrails are how you implement that constraint in code.

### What You Will Be Able to Do

By the end of the course you will be able to:

- Explain why a system prompt is a *request*, not an enforced boundary.
- Build a deterministic, explainable input filter and prove that blocked prompts never reach the model.
- Detect and redact PII in inbound traffic using a local validator, with no data leaving the machine.
- Judge when a guard should **block** (malicious intent) versus **redact and continue** (benign intent).
- Evaluate a third-party validator before depending on it, including whether it still installs.
- Enforce structured JSON output and tune toxicity filtering with a calibrated threshold.
- Check output grounding with embeddings, and let the pipeline retry or fall back instead of shipping unverified text.
- Assemble input and output guardrails into one Pydantic AI agent, regression-tested with a deterministic resiliency harness.

## Before You Start

### 1. Dependencies

This project uses [uv](https://docs.astral.sh/uv/). All modules share one dependency set, so install once:

```bash
uv sync
```

> **Do not `pip install` individual packages.** If something is missing, add it to `pyproject.toml`
> and re-run `uv sync`. The modules are written against the pinned versions in `uv.lock`.

### 2. Credentials

Copy the template and fill in your own values:

```powershell
Copy-Item .env.example .env      # Windows PowerShell
cp .env.example .env             # macOS / Linux
```

```ini
USE_OLLAMA=false                 # true to run fully locally against Ollama
LLM_MODEL=gpt-4o-mini            # the BARE model id - no "openai:" prefix
API_KEY=sk-...                   # not needed when USE_OLLAMA=true
BASE_URL=                        # blank for OpenAI; set for other endpoints
```

`.env` is gitignored. **Never commit a real key.**

### 3. Run Both Backends

Every module is written to work against **OpenAI or a local Ollama server** by changing one variable.
Ollama is worth trying: it is free, offline, and makes it obvious that none of the guard logic depends
on a particular provider.

| Variable | OpenAI | Ollama |
|----------|--------|--------|
| `USE_OLLAMA` | `false` | `true` |
| `LLM_MODEL` | `gpt-4o-mini` | `llama3` (whatever you pulled) |
| `API_KEY` | your key | any placeholder |
| `BASE_URL` | blank | `http://localhost:11434/v1` |

> **No credentials? You can still start.** Modules 1 and 3 include deterministic exercises that run
> fully offline. Module 3's guard tests, attack battery, and PII redaction all work with no API key at all.

## How the Course Is Organised

Each module follows the same shape, so you always know where you are:

| Section | Purpose |
|---------|---------|
| **Concept** | The idea being introduced, with tables and diagrams you can read without running anything |
| **Runnable Lab** | Code cells that demonstrate the concept and print what happened |
| **Why It Matters** | The engineering consequence, including honest limits of the technique |
| **Summary & Assignment** | What you should take away, plus exercises to push further |

### A Note on the Labs

The labs are deliberately **honest about limits**. Where a detector only catches known attack shapes, the
module says so rather than implying it is a security control. Where a Hub validator is retired or cannot
install, the module reports that instead of quietly dropping the example.

That matters because the failure mode in this field is false confidence. A regex that always reports
"safe" is worse than no check at all, since it stops you looking.

> **The detectors in these labs are teaching aids.** Real guardrails run in trusted application code
> *around* the model - which is exactly what you build from Module 3 onward.

### Troubleshooting

| Symptom | Cause | Fix |
|---------|-------|-----|
| `model not found` / 404 from OpenAI | Model id has an `openai:` prefix | Set the bare id in `.env`, e.g. `gpt-4o-mini` |
| `Connection refused` with `USE_OLLAMA=true` | Ollama is not running | Run `ollama serve`, then `ollama pull <model>` |
| Garbled `âœ“` characters in old output | Non-UTF-8 console outside Jupyter | Harmless; re-run the cell. Jupyter renders UTF-8 correctly |
| Slow first PII detection | Presidio downloads the spaCy `en_core_web_lg` model (~400 MB) | One-time cost; pattern entities work without it |
| Guardrails telemetry warnings about a network host | The library tries to phone home | Ignore, or disable Guardrails telemetry |
